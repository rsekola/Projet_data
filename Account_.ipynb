```python
class Compte:
    def __init__(self, numero_de_compte: str, solde_du_compte: float, titulaire_du_compte: str):
        self.numero_de_compte = numero_de_compte
        self.solde_du_compte = solde_du_compte
        self.titulaire_du_compte = titulaire_du_compte

    def afficher_informations(self):
        print(f"Numéro de compte : {self.numero_de_compte}")
        print(f"Titulaire du compte : {self.titulaire_du_compte}")
        print(f"Solde du compte : {self.solde_du_compte:} ")
```

```python
def deposer(self, montant: float):
        if montant > 0:
            self.solde_du_compte += montant
            print(f"{montant:}")
        else:
            print("Dépôt doit être positif")
```

```python
def retirer(self, montant: float):
        if montant <= 0:
            print("Retrait doit être positif")
        elif montant > self.solde_du_compte:
            print("Solde insuffisant pour effectuer le retrait")
        else:
            self.solde_du_compte -= montant
            print(f"{montant:.f}")
```

```python
  def check_balance(self):
        return self.solde_du_compte
```

```python
my_account = Compte("CI0156789", 450000.0, "Sekola Romaric")
compte2 = Compte("SN987654321", 300000.0, "Aminata Diop")
compte3 = Compte("BF123456789", 150000.0, "Oumar Traoré")
compte4= Compte("AY123456789", 500000.0, "Franck Zago")

my_account.deposer(50000)
my_account.retirer(100000)
print(f"Solde actuel de my_account : {my_account.check_balance():} ")

compte2.deposer(20000)
compte2.retirer(5000)
print(f"Solde actuel de compte2 : {compte2.check_balance():}")

compte3.retirer(20000)
compte3.retirer(200000)
compte3.deposer(50000)
print(f"Solde actuel de compte3 : {compte3.check_balance():}")

compte4.deposer(20000)
compte4.retirer(5000)
print(f"Solde actuel de compte4 : {compte4.check_balance():}")

my_account.afficher_informations()
compte2.afficher_informations()
compte3.afficher_informations()
compte4.afficher_informations()
```
