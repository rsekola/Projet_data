Plus le glucose est élevé → plus le risque de diabète augmente.

Le vieillissement et le surpoids renforcent la probabilité d’un diagnostic positif.

Certaines variables nécessitent un nettoyage préalable pour améliorer la fiabilité.

La glycémie est le facteur le plus déterminant du diabète.

Le surpoids et l’âge renforcent la probabilité de diagnostic.

Des incohérences existent dans certaines colonnes médicales, nécessitant une phase de nettoyage avant toute modélisation.

✅ 1. Gérer correctement les valeurs manquantes déguisées

Plusieurs variables critiques contiennent des zéros irréalistes (Glucose, Insulin, BMI, BloodPressure, SkinThickness).

Actions recommandées :

Remplacer les zéros par NaN

Appliquer ensuite :

✅ Imputation par la médiane (plus robuste que la moyenne)

ou

✅ Suppression des lignes si elles sont trop nombreuses et biaisantes

✅ 2. Traiter les valeurs aberrantes

Certaines colonnes comme Insulin et SkinThickness présentent des valeurs extrêmes.

Méthodes conseillées :

Utiliser l’IQR (Interquartile Range)

Visualisation avec boxplots

Winsorisation ou normalisation si nécessaire

✅ 3. Normalisation / Mise à l’échelle

Avant toute modélisation :

Appliquer StandardScaler ou MinMaxScaler

Améliore la performance des modèles comme :

Régression logistique

SVM

KNN

✅ 4. Modélisation prédictive

Objectif : prédire la variable Outcome

Modèles suggérés :

📌 Régression Logistique (baseline)

🌳 Random Forest

⚡ XGBoost / Gradient Boosting

🤖 K-Nearest Neighbors

Étapes :

Séparer les données (80% train / 20% test)

Entraîner le modèle

Évaluer via :

Accuracy

Precision / Recall

Matrice de confusion

ROC-AUC

✅ 5. Sélection de variables

Identifier les variables les plus influentes :

Importance des features

Corrélation

Analyse SHAP pour interprétabilité
