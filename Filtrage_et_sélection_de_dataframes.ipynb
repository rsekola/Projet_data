```python
import pandas as pd

# Création du dataframe
data = {
    'Name': ['John', 'Mary', 'Bob', 'Sarah', 'Tom', 'Lisa'],
    'Department': ['IT', 'Marketing', 'Sales', 'IT', 'Finance', 'Marketing'],
    'Age': [30, 40, 25, 35, 45, 28],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Salary': [50000, 60000, 45000, 55000, 70000, 55000],
    'Experience': [3, 7, 2, 5, 10, 4]
}
employee_df = pd.DataFrame(data)
```

```python
# Sélectionner les 3 premières lignes avec iloc
first_three_rows = employee_df.iloc[:3]
print(first_three_rows, "\n")
```

```python
# Sélectionner les lignes où le département est "Marketing" avec loc
marketing_rows = employee_df.loc[employee_df['Department'] == 'Marketing']
print(marketing_rows, "\n")
```

```python
# Sélectionner les colonnes Age et Gender pour les 4 premières lignes avec iloc
age_gender_first_four = employee_df.iloc[:4, [2, 3]]
print(age_gender_first_four, "\n")
```

```python
# Sélectionner Salary et Experience pour les lignes où Gender = 'Male' avec loc
salary_exp_male = employee_df.loc[employee_df['Gender'] == 'Male', ['Salary', 'Experience']]
print("4. Salary et Experience pour les employés masculins :")
print(salary_exp_male)
```
