```python
import numpy as np
file = open("Loan_prediction_dataset.csv", "r")
data = np.genfromtxt(file, delimiter=",", skip_header=1, usecols=7)
file.close()
moyenne = np.mean(data)
mediane = np.median(data)
ecart_type = np.std(data)

print(f"{moyenne}")
print(f"{mediane}")
print(f"{ecart_type}")
```
