```python
import os
import pandas as pd

# Définir le répertoire de sauvegarde
repertoire_sauvegarde = '/content/mon_dossier'

# Créer le répertoire s'il n'existe pas
os.makedirs(repertoire_sauvegarde, exist_ok=True)

# Liste des cibles
targets = [25300, 25286, 25287, 25257, 25308, 25286, 25275, 25274]

# Copier le DataFrame (assurez-vous que df est déjà défini dans votre environnement)
df = df.copy()

# Liste pour stocker les fichiers
files = []
current_index = 0

# Découper le DataFrame en fonction des cibles de poids
for t in targets:
    temp = []
    total = 0

    while total < t and current_index < len(df):
        row = df.iloc[current_index].copy()
        w = row["Net Weight (KG)"]

        if total + w <= t:
            temp.append(row)
            total += w
            current_index += 1
        else:
            # Séparer le poids
            diff = t - total
            row["Net Weight (KG)"] = diff
            temp.append(row)

            # Le poids restant va dans la ligne suivante
            df.at[current_index, "Net Weight (KG)"] = w - diff
            total += diff

    files.append(pd.DataFrame(temp))

# Sauvegarder les fichiers dans le répertoire créé
paths = []
for i, f in enumerate(files, 1):
    chemin_fichier = os.path.join(repertoire_sauvegarde, f"Net_Weight_split_{i}.xlsx")
    f.to_excel(chemin_fichier, index=False)
    paths.append(chemin_fichier)

# Afficher les chemins des fichiers sauvegardés
print(paths)
```
