```python
import math

class Calculatrice:
    def __init__(self):

        self.operations = {
            '+': lambda x, y: x + y,
            '-': lambda x, y: x - y,
            '*': lambda x, y: x * y,
            '/': lambda x, y: x / y
        }

    def add_operation(self, symbole, fonction):

        self.operations[symbole] = fonction

    def calculer(self, symbole, *operandes):

        if symbole in self.operations:
            try:
                return self.operations[symbole](*operandes)
            except TypeError:
                return "Erreur : nombre d’arguments invalide"
        else:
            return "Erreur : opération inconnue"


def exponentiation(x, y):
    return x ** y

def racine_carree(x):
    return math.sqrt(x)

def logarithme(x, base=math.e):
    return math.log(x, base)


if __name__ == "__main__":
    calc = Calculatrice()


    calc.add_operation('^', exponentiation)
    calc.add_operation('√', racine_carree)
    calc.add_operation('log', logarithme)

    print("=== Calculatrice Interactive ===")
    print("Opérations disponibles : +, -, *, /, ^, √, log")
    print("Tapez 'q' pour quitter.\n")

    while True:
        symbole = input("Entrez une opération (+, -, *, /, ^, √, log ou q pour quitter) : ").strip()

        if symbole.lower() == 'q':
            print("Merci d’avoir utilisé la calculatrice. À bientôt !")
            break


        if symbole not in calc.operations:
            print("Opération inconnue ! Essayez encore.\n")
            continue

        try:
            if symbole == '√':
                x = float(input("Entrez le nombre : "))
                resultat = calc.calculer(symbole, x)
            elif symbole == 'log':
                x = float(input("Entrez le nombre : "))
                base = input("Entrez la base (par défaut e) : ").strip()
                base = float(base) if base else math.e
                resultat = calc.calculer(symbole, x, base)
            else:
                x = float(input("Entrez le premier nombre : "))
                y = float(input("Entrez le second nombre : "))
                resultat = calc.calculer(symbole, x, y)

            print(f"Résultat = {resultat}\n")

        except ValueError:
            print("Entrée invalide, veuillez entrer des nombres valides.\n")
```
