```python
from ydata_profiling import ProfileReport
import pandas as pd

df = pd.read_csv("Tunisair_flights_dataset.csv")

profile = ProfileReport(
    df,
    title="Rapport ydata-Profiling - Tunisair Flights Dataset",
    explorative=True
)

profile.to_file("rapport_ydata_profiling.html")
```

Après analyse du rapport généré par ydata-profiling, aucune valeur manquante n’a été détectée dans le jeu de données

Le rapport ydata-profiling montre que le jeu de données Tunisair contient une seule variable numérique mesurable directement : Arrival delay (retard à l’arrivée)

D’après le rapport généré par ydata-profiling, les valeurs aberrantes concernent principalement la variable numérique : Arrival delay (Retard à l’arrivée)

L’exploration du jeu de données Tunisair avec pandas et ydata-profiling montre que les données sont de bonne qualité, sans valeurs manquantes et correctement structurées. La majorité des variables sont catégorielles, avec une seule variable numérique : le retard à l’arrivée.

Aucune corrélation significative n’a été observée, en raison du faible nombre de variables numériques. Toutefois, une tendance claire apparaît : la plupart des vols ont peu ou pas de retard, tandis qu’un petit nombre présente des retards très élevés, considérés comme des valeurs aberrantes.
