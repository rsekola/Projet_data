```python
import numpy as np
grades = np.array([85, 90, 88, 92, 95, 80, 75, 98, 89, 83])
```

```python
mean_grade = np.mean(grades)
median_grade = np.median(grades)
std_dev = np.std(grades)
```

```python
max_grade = np.max(grades)
min_grade = np.min(grades)
```

```python
sorted_grades = np.sort(grades)
```

```python
index_max = np.argmax(grades)
```

```python
count_above_90 = np.sum(grades > 90)
```

```python
percent_above_90 = np.mean(grades > 90) * 100
```

```python
percent_below_75 = np.mean(grades < 75) * 100
```

```python
high_performers = grades[grades > 90]
```

```python
passing_grades = grades[grades > 75]
```

```python
print("Notes :", grades)
print("Moyenne :", mean_grade)
print("Médiane :", median_grade)
print("Écart type :", std_dev)
print("Note maximale :", max_grade)
print("Note minimale :", min_grade)
print("Notes triées :", sorted_grades)
print("Indice de la note la plus élevée :", index_max)
print("Nombre d’étudiants > 90 :", count_above_90)
print("Pourcentage d’étudiants > 90 :", percent_above_90, "%")
print("Pourcentage d’étudiants < 75 :", percent_below_75, "%")
print("High performers (notes > 90) :", high_performers)
print("Passing grades (notes > 75) :", passing_grades)
```
