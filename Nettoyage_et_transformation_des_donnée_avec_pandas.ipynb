```python
import pandas as pd

produits = ["Pomme","Mange","Ananas"]
series_produits = pd.Series(produits)
print(series_produits)

Prix =[6,8,9]
series_prix = pd.Series(Prix)
print(series_prix)

addition_series = series_produits + series_prix.astype(str)
print(addition_series)

df_dict = pd.DataFrame({"Produits":series_produits,"Prix":series_prix})
print(df_dict)

data_list = [
    {"Product": series_produits[i], "Price": series_prix[i]}
    for i in range(len(series_produits))
]

df_list = pd.DataFrame(data_list)
print(df_list)

```
